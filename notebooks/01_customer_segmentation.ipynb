```python
earliest_date = sales["order_date"].min()
latest_date = sales["order_date"].max()
print(earliest_date, latest_date)
```

```python
sales.dropna(subset = ["order_date"], inplace = True)
```

```python
display(sales.isna().sum())
display(sales.shape)
```

```python
cutoff = pd.Timestamp("2013-07-01")
window_end = cutoff + pd.DateOffset(months = 6)
```

```python
history = sales[sales["order_date"] < cutoff]
future = sales[sales["order_date"].between(cutoff, window_end, inclusive = "left")]

print(history.shape, future.shape)
```

```python
history_customers = history["customer_key"].drop_duplicates()
print(len(history_customers))
```

```python
came_back = history_customers.isin(future["customer_key"])
print(came_back.head())
```

```python
n_customers = len(history_customers)
n_came_back = came_back.sum()
churn_rate = 1 - n_came_back / n_customers

print("Customers in history:", n_customers)
print("Came back:", n_came_back)
print("Churn rate:", round(churn_rate * 100, 1), "%")
```

```python
features = history.groupby("customer_key").agg(
    frequency = ("order_number", "nunique"),
    monetary = ("sales", "sum"),
    last_order = ("order_date", "max"),
    first_order = ("order_date", "min")
)
```

```python
display(features.head())
display(features.shape)
display(features.describe())
```

```python
features["recency"] = (cutoff - features["last_order"]).dt.days
features["tenure"] = (cutoff - features["first_order"]).dt.days
```

```python
display(features["recency"].head())
display(features["recency"].shape)
display(features["recency"].describe())

display(features["tenure"].shape)
display(features["tenure"].head())
display(features["tenure"].describe())
```

```python
print((features["tenure"] >= features["recency"]).all())
```

```python
features = features.drop(columns = ["last_order", "first_order"])
features.head()
```

```python
history_cat = history.merge(products[["product_key", "category"]], on = "product_key", how = "left")
```

```python
history_cat.head()
```

```python
display(history_cat.shape)
display(history_cat["category"].isna().sum())
```

```python
check = ((history_cat["category"] == "Accessories").sum()) + ((history_cat["category"] == "Clothing").sum()) + ((history_cat["category"] == "Bikes").sum())

print(check, history_cat.shape[0])
```

```python
history_cat.head()
```

```python
history_cat["is_bike"] = history_cat["category"] == "Bikes"
history_cat["is_clothing"] = history_cat["category"] == "Clothing"
history_cat["is_accessories"] = history_cat["category"] == "Accessories"
history_cat.head()
```

```python
category_counts = history_cat.groupby("customer_key").agg(
    n_bikes = ("is_bike", "sum"),
    n_clothing = ("is_clothing", "sum"),
    n_accessories = ("is_accessories", "sum")
)

display(category_counts.head())
display(category_counts.shape)
```

```python
features = features.join(category_counts, how = "left")

display(features.head())
display(features.isna().sum())
```
